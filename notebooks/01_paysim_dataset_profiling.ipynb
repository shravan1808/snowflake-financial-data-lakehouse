```python
from pathlib import Path

output_dir = Path('D:/Colaberry/KMIT/Week_06/snowflake-financial-data-lakehouse/snowflake-financial-data-lakehouse/batches')
output_dir.mkdir(exist_ok=True)

min_range = paysim_data['step'].min()
max_range = paysim_data['step'].max()
step_ranges = []
for start in range(min_range,max_range,100):
    if start==301:
        end = max_range
        step_ranges.append([start,end])
        break
    else:
        end = min(start+99,max_range)
        step_ranges.append([start,end])

for i,batch in enumerate(step_ranges,start=1):
    filename = f"batch_{i:03d}.csv"
    start = batch[0]
    end = batch[1]
    data = paysim_data[(paysim_data['step']>=start) & (paysim_data['step']<=end)]
    data.to_csv(output_dir/filename,index=False)
```
